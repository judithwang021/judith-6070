```python
letters = ['a',  'b',  'c', 'd']
letters[int(int('3'*2)//11)]
```

letters[int(int('3'*2)//11)] evaluate to letter[3], which is the fourth letter 'd'.

```python
letters[-1]
```

letters[-1] evaluates to the last letter in the list, which is 'd'.

```python
letters[:2]
```

letters[:2] evaluates to the first two letters in the list.
