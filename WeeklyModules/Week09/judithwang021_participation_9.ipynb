```python
import os, requests, bs4, re
import requests, bs4
from IPython.display import Image, display
url = 'https://sp2.upenn.edu/sp2-directory/?_departments=faculty'

response = requests.get(url)
response.raise_for_status()
soup = bs4.BeautifulSoup(response.content, 'html.parser')
print(soup.title.get_text(strip=True))
```

```python
image_tags = soup.find_all('img', src=True)

print('Number of image tags:', len(image_tags))

for image in image_tags[:15]:
    print('class:', image.get('class'))
    print('alt:', image.get('alt'))
    print('src:', image.get('src'))
    print()
```

```python
first_photo = soup.find('img', class_='card-image')

parent = first_photo.parent

for number in range(15):
    print('Layer:', number)
    print('Tag:', parent.name)
    print('Class:', parent.get('class'))
    print()

    parent = parent.parent
```

```python
cards = soup.find_all('article', class_='card--sp2_directory')

print('Number of cards:', len(cards))

for card in cards[:15]:
    print(card.get_text(' ', strip=True))
    print()
```

```python
photo_tags = soup.find_all('img', class_='card-image')

for photo in photo_tags:
    photo_url = photo.get('src')
    display(Image(url=photo_url, width=180))
```

```python
folder_name = 'sp2_faculty_photos'
os.makedirs(folder_name, exist_ok=True)

photo_tags = soup.find_all('img', class_='card-image')

for number, photo in enumerate(photo_tags, start=1):
    photo_url = photo.get('src')

    image_response = requests.get(photo_url)
    image_response.raise_for_status()

    file_name = f'faculty_{number}.jpg'
    file_path = os.path.join(folder_name, file_name)

    with open(file_path, 'wb') as image_file:
        image_file.write(image_response.content)
```
