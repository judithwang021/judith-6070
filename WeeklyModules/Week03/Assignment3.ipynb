```python
def collatz(number):
  if number % 2 == 0:
    result = number // 2
  else:
    result = 3 * number + 1
  print(result, end=' ')
  return result

response = input('Enter a starting number (greater than 0): ')
n = int(response)
print(n, end=' ')

while n != 1:
  n = collatz(n)
```
