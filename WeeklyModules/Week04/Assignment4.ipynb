```python
ring_of_fire = [
    'Bolivia',
    'Chile',
    'Ecuador',
    'Peru',
    'Costa Rica',
    'Guatemala',
    'Mexico',
    'Japan',
    'Philippines',
    'Indonesia',
    'New Zealand',
    'Canada',
    'Fiji',
    'Alaska',
    'Washington',
    'California',
    'Russia',
    'Taiwan',
    'Tonga',
    'Kermadec Islands']
len(ring_of_fire)
```

```python
antarctic = parsed['parsed_place'].str.contains('Antarctic', na=False)
antarctic
parsed.loc[antarctic, 'parsed_place'].value_counts()
```

```python
parsed['ring_of_fire'] = (parsed['parsed_place'].isin(ring_of_fire) |
                          parsed['parsed_place'].str.contains('Antarctic', na=False))
parsed['ring_of_fire'].value_counts()
```

```python
inside = parsed['ring_of_fire'].sum()
outside = (~parsed['ring_of_fire']).sum()
print(inside)
print(outside)
```
